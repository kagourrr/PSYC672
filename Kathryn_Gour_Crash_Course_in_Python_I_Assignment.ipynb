## 10) User-defined Functions
Good experimental practice involves the standardization of research materials.  An example is when the materials involve visual stimuli and it is important to standardize the experience of participants.  This involves not only presenting the exact same stimulus but also how the participant sees the stimulus.  For example, presenting identical images to two participants will result in different experiences if they are looking at the same monitor, but one of the participants is positioned at a distance of 2 feet from the image and the other is positioned at a distance of 10 feet.  For this reason, visual stimuli are often measured in terms of “visual angle”, the distance an object subtends at the eye. Usually measured in degrees of arc, the visual angle ($V$) can be computed as $V=2*arctan(\frac{S}{2D}) $, where $S$ = the span (width) of the stimulus and $D$ = the distance from the viewer to the stimulus.  

In the code block below:

1) Use the “def” expression to write your own function named “Visual_Angle”. The function should take two values as inputs, the span of the stimulus and the distance from the viewer to the stimulus. The function should return the visual angle (V) as output. (HINT: you may need to import the math module).

2) Demonstrate the use of your new function in a code cell using at least two sets of values for S and D.


```python
import math
```

```python
def VisAng(S,D):
  V = 2*math.atan(S/(2*D))
  return V
```

## 14) Writing your own data collection program.
 Putting together the concepts you have learned thus far, use the code block below to write your own data collection function that will collect information from a participant in a research study. The program should ask users to input their: <br>
1) Name <br>
2) Gender<br>
3) Age<br>
4) A list of any medications taken in the past 24 hours (separated by commas). <br>
5) Responses to a brief (5-10 questions) questionnaire of your chosing (e.g., depression, anxiety, etc.)
      <p>
  
**REQUIRED BEHAVIORS OF YOUR DATA COLLECTION FUNCTION**<br>
In addition to simply collecting information from the user, it is important to ensure the integrity of the data collected. Thus, the program should perform some quality control along the way. For example, the program should be designed so that:<br>
1)   Only numeric responses are accepted in response to the question about age.  If the user enters a non-numeric response, the program should instruct the user to enter a number and not continue until a numeric input is provided.<br>
2)   If a participant is less than 18 years old, the program should halt and print a message to the user that they are not of legal age to provide informed consent and will not be able to participate in your study. Also, if the age entered is greater than 100, the program should prompt the user to confirm their age and/or change their age before continuing.<br>
3)   The program should also check the medications input by the user to see if either Thorazine or Haldol were recently taken. In the event that either of those medications are reported, the program should halt and inform the user that he/she is not eligible to participate in the study (keep in mind that some users may capitalize the drug names while others may not; HINT: the str.split() method may come in handy here). <br>
4) The program should also randomly assign the participant an identification number and print that number to the screen after all data have been entered.<br>
5) Finally the program should return, when finished running, a list object containing the name, gender, age, participant identifier, medication data, and responses to the questionnaire.


```python
import random
```

```python
def participant_data():
  print('Welcome to the CPL! Today you will be participating in a study examining anxiety. Please follow the directions and enter your responses as indicated.')
  name = input('What is your name? ')
  gender = input('What is your gender? ')

  #let's check age!
  while True:
      age = input('What is your age?')
      if not age.isdigit():
        print('Please enter a number.')
        continue
      participant_age = int(age)

      if participant_age < 18:
        print('Thank you for your participation. Unfortunately, you are not of legal age to participate in this study.')
        return None
      elif participant_age > 100:
          confirm = input(f'You entered {participant_age}. Is this correct? (Y/N):')
          if confirm.upper() == 'N':
            print('Please re-enter your age.')
            continue
      break

  meds = input('List any medications taken in the past 24 hours. Make sure you separate your answers using commas:')
  participant_meds = [m.strip().lower() for m in meds.split(',')]

  if 'thorazine' in participant_meds or 'haldol' in participant_meds:
    print('Thank you for your participation. Unfortunately, you are not eligible to participate in this study.')
    return None

  #Using GAD-7 bc short lol
  questionnaire = []
  print('\nPlease respond to the following questions on a scale from 1 (Not at all) to 5 (Nearly every day):')
  questions = [
      'Feeling nervous, anxious, or on edge.',
      'Not being able to stop or control worrying.',
      'Worrying too much about different things.',
      'Trouble relaxing.',
      'Being so restless that it is hard to sit still.',
      'Becoming easily annoyed or irritable.',
      'Feeling afraid, as if something awful might happen.'
    ]

  for q in questions:
    while True:
      resp = input(q + '')
      if resp.isdigit() and 1 <= int(resp) <=5:
        questionnaire.append(int(resp))
        break
      else:
        print('Please enter a number between 1 and 5.')

  #Time for PID
  PID = random.randint(1000, 9999)
  print(f'\nYour participant ID is: {PID}.')

  data = [name, gender, participant_age, PID, participant_meds, questionnaire]
  return data
```
